```python
from dotenv import load_dotenv

import os
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
#Langsmith Tracking
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
#Langchain Tracing
os.environ["LANGCHAIN_TRACING_V2"] = "true"

os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")#Langsmith Tracing


from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-120b",api_key=groq_api_key
)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="Translate the following from English to French"),
    HumanMessage(content="Hello how are you?")
]

response = model.invoke(messages)
```

```python
type(messages[0])
```

```python
response
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

parser.invoke(response)
```

```python
##Using LCEL - chain the components

chain = model|parser
chain.invoke(messages)
```

```python
##Prompt Template

from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following into {language}"
prompt = ChatPromptTemplate.from_messages(
    [
        ("system",generic_template),
        ("user","{text}")
    ]
)
```

```python
result = prompt.invoke({"language":"Telugu","text":"Hello how are you?"})
```

```python
result.to_messages()
```

```python
##chaining together components with LCEL
chain = prompt|model|parser

chain.invoke({"language":"Telugu","text":"Hello how are you?"})
```
