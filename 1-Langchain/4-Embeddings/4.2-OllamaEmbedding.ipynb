### Ollama

Ollama supports embedding models, making it possible to build retrieval augmented generation(RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings

embedding = OllamaEmbeddings(model="nomic-embed-text") ##by default llama2 model will be chosen, explicitly mention other models to use
embedding
```

```python
r1 = embedding.embed_documents(
    [
        "Alpha is the first letter of Greek Language",
        "Beta is the second letter of Greek Language"
    ]
)
```

```python
embedding.embed_query("What is the second letter of Greek Language")
```

```python
## other ollama embedding models

embedding = OllamaEmbeddings(model="mxbai-embed-large") ##by default llama2 model will be chosen, explicitly mention other models to use
embedding
```

```python
text = "What is the second letter of Greek Language"
result = embedding.embed_query(text)
result
```
