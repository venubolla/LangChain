```python
from dotenv import load_dotenv

import os
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
#Langsmith Tracking
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
#Langchain Tracing
os.environ["LANGCHAIN_TRACING_V2"] = "true"

os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")#Langsmith Tracing


from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b"
)

print(llm)
```

```python
response = llm.invoke("Hello, who are you?")
print(response.content)
```

```python
response = llm.invoke("What is Generative AI?")
print(response.content)
```

```python
### ChatPrompt Template
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are an experst AI Engineer. Provide me answers based on the question"),
        ("user","{input}")
    ]
)

prompt
```

```python
chain = prompt|llm 

# | sign is basically saying that we need to combine the prompt with llm that actually
# becomes a chain

response = chain.invoke({"input":"Can you tell me about Langsmith?"})
print(response)
```

```python
type(response)
```

```python
## stroutput Parser

from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt|llm|output_parser
response = chain.invoke({"input":"Can you tell me about Langsmith?"})
print(response)
```
