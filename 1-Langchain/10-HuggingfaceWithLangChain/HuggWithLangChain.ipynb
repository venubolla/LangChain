```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint, ChatHuggingFace
```

```python
hf_token = os.getenv("HF_TOKEN") or os.getenv("HUGGINGFACEHUB_API_TOKEN")



if not hf_token:

    raise ValueError("Set HF_TOKEN or HUGGINGFACEHUB_API_TOKEN in your .env file.")



llm = HuggingFaceEndpoint(

    repo_id="meta-llama/Llama-3.1-8B-Instruct",

    task="conversational",

    provider="auto",

    max_new_tokens=256,

    temperature=0.3,

    huggingfacehub_api_token=hf_token,

)

chat_model = ChatHuggingFace(llm=llm)
```

```python
chat_model.invoke("What is Machine Learning?")
```

```python
from langchain_core.prompts import PromptTemplate



template = """

Question: {question}



Answer: Let's think step by step.

"""



prompt = PromptTemplate.from_template(template)

print(prompt)
```

```python
llm_chain = prompt | chat_model



response = llm_chain.invoke({"question": "What is the capital of France?"})

response.content
```
